# Feature Engineering and Ploynomial Regression

````python
import numpy as np
import matplotlib.pyplot as plt
from lab_utils_multi import zscore_normalize_features, run_gradient_descent_feng
np.set_printoptions(precision=2)
````

## Ploynomial Features

````python
x = np.arange(0, 20, 1)
y = 1 + x**2
X = x.reshape(-1, 1)
````

````python
model_w,model_b = run_gradient_descent_feng(X,y,iterations=1000, alpha = 1e-2)

plt.scatter(x, y, marker='x', c='r', label='Actual Value')
plt.title('No Feature Engineering')
plt.plot(x, X@model_w + model_b, label='Predicted Value')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.show()
````

modify input data to engineer the needed features

````python
# create target data
x = np.arange(0, 20, 1)
y = 1 + x**2
X = x**2 
````

````python
X = X.reshape(-1, 1)
model_w,model_b = run_gradient_descent_feng(X, y, iterations=10000, alpha = 1e-5)

plt.scatter(x, y, marker='x', c='r', label='Actual Value')
plt.plot(x, np.dot(X, model_w) + model_b, label='Predicted Value')
plt.title('Engineered Features')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
````

# Selecting Features
one could add a variety of potential features to try and find the most useful <br>
--> try $y=w_0x_0 + w_1x_1^2+w_2x_2^3+b$

```np.c_[]``` concatenates columns

````python
# create target data
x = np.arange(0, 20, 1)
y = x**2

# engineer features
X = np.c_[x, x**2, x**3]
````

````python
model_w,model_b = run_gradient_descent_feng(X, y, iterations=10000, alpha=1e-7)

plt.scatter(x, y, marker='x', c='r', label="Actual Value")
plt.title("x, x**2, x***3 features")
plt.plot(x, np.dot(X, model_w) + model_b, label='Predicted Value')
plt.legend()
plt.show()
````

### Scaling Features
apply z-score normalization to our examples

````python
# create target data
x = np.arange(0,20,1)
X = np.c_[x, x**2, x**3]
print(f"Peak to Peak range by column in Raw        X:{np.ptp(X,axis=0)}")

# add mean_normalization 
X = zscore_normalize_features(X)     
print(f"Peak to Peak range by column in Normalized X:{np.ptp(X,axis=0)}")
````

````python
x = np.arange(0,20,1)
y = x**2

X = np.c_[x, x**2, x**3]
X = zscore_normalize_features(X) 

model_w, model_b = run_gradient_descent_feng(X, y, iterations=100000, alpha=1e-1)

plt.scatter(x, y, marker='x', c='r', label="Actual Value"); plt.title("Normalized x x**2, x**3 feature")
plt.plot(x,X@model_w + model_b, label="Predicted Value"); plt.xlabel("x"); plt.ylabel("y"); plt.legend(); plt.show()
````
