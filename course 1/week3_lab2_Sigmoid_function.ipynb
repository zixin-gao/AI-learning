## Sigmoid / Logistic Function
- we would like the predictions of our classification model to be between 0 and 1
- this can be accomplished by using a sigmoid function which maps all input values between 0 and 1

The general formula for a sigmoid function is: <br>
$$ g(z) = \frac{1}{1 + e^{-z}} $$

- z is the output of a linear regression model
- in the case of a single example, $z$ is scalar
- in the case of multiple examples, $z$ may be a vector consisting of $m$ values, one for each example
- the implementation should cover both of these potential input formats

Numpy has a function called ```exp()```, which offers a way to calculate the exponential of all elements in the input array ```z```

````python
# input is an array
input_array = np.array([1, 2, 3])
exp_array = np.exp(input_array)

input_array
exp_array

# also works for single number
````

````python
def sigmoid(z):
    """
    Compute the sigmoid of z
    Args:
        z (ndarray): A scalar, numpy array of any size

    Returns:
        g (ndarray): sigmoid(z), same shape as z
    """
    return 1 / (1 + np.exp(-z))
````

let's see what the output of this function is for various values of z


````python
# Generate an array of evenly spaced values between -10 and 10
z_tmp = np.arange(-10, 11)
y = sigmoid(z_tmp)

# code for pretty printing the two values beside one another
np.set_printoptions(precision=3)
print("Input (z), Output (sigmoid(z))")
print(np.c_[z_tmp, y])
````

````python
# plot in matplotlib
fig, ax = plt.subplots(1, 1, figsize=(5, 3))
ax.plot(z_tmp, y, c='b')
ax.set_xlabel('z')
ax.set_ylabel('y')
ax.set_title('Sigmoid function')
draw_vthresh(ax, 0)
````
